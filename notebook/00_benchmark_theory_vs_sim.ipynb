# 00 — Benchmark theory vs simulation

Reads `result/table/00_benchmark_theory_vs_sim.csv` and saves a vector PDF to `result/figure/00_benchmark_theory_vs_sim.pdf`.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path("..").resolve()
csv_path = ROOT / "result" / "table" / "00_benchmark_theory_vs_sim.csv"
fig_path = ROOT / "result" / "figure" / "00_benchmark_theory_vs_sim.pdf"

df = pd.read_csv(csv_path)
df = df.sort_values("K")

plt.rcParams["pdf.fonttype"] = 42
plt.figure(figsize=(7,4))
plt.plot(df["K"], df["theory_E_sumTopB_eta"], label="Theory (asymptotic line)")
plt.errorbar(df["K"], df["sim_E_sumTopB_eta"], yerr=2*df["sim_SE"], fmt="o", capsize=3, label="Simulation (mean ± 2 SE)")
plt.xlabel("K (inspections)")
plt.ylabel("E[ Σ_{top-B} η ]")
plt.title(f"Benchmark theory vs simulation (p={df.p.iloc[0]}, μ={df.mu.iloc[0]}, α={df.alpha.iloc[0]})")
plt.legend()
plt.tight_layout()
fig_path.parent.mkdir(parents=True, exist_ok=True)
plt.savefig(fig_path, format="pdf")
plt.show()

print("Saved:", fig_path)
```
