# 02 — Lemma 4 enrichment bound

Reads `result/table/02_lemma4_enrichment_bound.csv` and saves a vector PDF to `result/figure/02_lemma4_enrichment_bound.pdf`.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path("..").resolve()
csv_path = ROOT / "result" / "table" / "02_lemma4_enrichment_bound.csv"
fig_path = ROOT / "result" / "figure" / "02_lemma4_enrichment_bound.pdf"

df = pd.read_csv(csv_path).sort_values("alpha")

plt.rcParams["pdf.fonttype"] = 42
plt.figure(figsize=(7,4))
plt.plot(df["alpha"], df["lemma4_bound"], label="Lemma 4 bound")
plt.errorbar(df["alpha"], df["emp_P_T1_given_selected"], yerr=2*df["emp_SE"], fmt="o", capsize=3, label="Empirical (mean ± 2 SE)")
plt.xscale("log")
plt.xlabel("α = P(S=1) (log scale)")
plt.ylabel("P(T=1 | S=1)")
plt.title(f"Selection enrichment (p={df.p.iloc[0]}, μ={df.mu.iloc[0]}, J≈{df.J.iloc[0]:.3g})")
plt.legend()
plt.tight_layout()
fig_path.parent.mkdir(parents=True, exist_ok=True)
plt.savefig(fig_path, format="pdf")
plt.show()

print("Saved:", fig_path)
```
